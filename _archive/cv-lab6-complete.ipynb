# Задание 1. CNN локализация лицевых точек

```python
path = '/kaggle/input/datasets/slnkvdns/facial-keypoints-detection/training.csv'
df = pd.read_csv(path)
df.info()
```

```python
class FacialKeypointsDetectionDataset:
    def __init__(self, df, transforms, is_train):
        self.transforms = transforms
        if is_train:
            self.df = df.fillna(-1)
        else:
            self.df = df.fillna(-1)

    def __len__(self):
        return self.df.shape[0]

    def __getitem__(self, idx):
        target = torch.tensor(self.df.iloc[idx, :-1].to_numpy().astype('float32'))
        coords = target.reshape((-1, 2))
        
        mask = (coords != -1).unsqueeze(0)

        str_img = self.df.iloc[idx, -1]
        img = np.array(str_img.split()).astype('uint8').reshape(96, 96)
        img = np.stack([img] * 3, axis=-1)
        img = Image.fromarray(img)
        
        coords = tv_tensors.KeyPoints(coords.unsqueeze(0), canvas_size=(96, 96))

        transformed_img, transformed_coords = self.transforms(img, coords)
        transformed_coords /= transformed_img.shape[1]
        transformed_coords[~mask] = -1.0

        transformed_coords = transformed_coords.reshape(-1)

        return transformed_img, transformed_coords
```

```python
def draw_example(img, target):
    x = target[range(0, 30, 2)]
    y = target[range(1, 30, 2)]

    plt.imshow(img)
    for i in range(15):
        if x[i] != -1 and y[i] != -1:
            plt.scatter(x[i] * img.shape[1], y[i] * img.shape[1], c='red')

def dataset_img_to_normal_img(dataset_img):
    img = dataset_img.numpy().transpose(1, 2, 0)
    img = std * img + mean
    img = np.clip(img, 0, 1)
    return img
```

```python
def masked_rmse_loss(y_pred, y_true):
    mask = (y_true != -1)
    mse = ((y_pred - y_true) ** 2)[mask].sum() / (mask.sum() + 1e-8)
    rmse = torch.sqrt(mse + 1e-8) 
    return rmse

def masked_mse_loss(y_pred, y_true):
    mask = (y_true != -1)
    mse = ((y_pred - y_true) ** 2)[mask].sum() / (mask.sum() + 1e-8)
    return mse 
```

```python
def train_loop(model, loss, optimizer, dataloader, train_losses, train_rmse, scheduler=None, verbose=None):
    model.train()

    batch_size = dataloader.batch_size
    all_targets = []
    all_predictions = []

    is_xla = 'xla' in str(device)
    loader = pl.MpDeviceLoader(dataloader, device) if is_xla else dataloader


    if verbose is not None:
        ver = len(dataloader) / verbose
    else:
        ver = 10

    for batch, (X, y) in enumerate(loader):
        X, y = X.to(device), y.to(device)

        optimizer.zero_grad()
        y_pred = model(X)
        output = loss(y_pred, y)
        
        output.backward()
        
        if is_xla:
            xm.optimizer_step(optimizer)
        else:
            optimizer.step()

        all_predictions.append(y_pred.detach().cpu())
        all_targets.append(y.detach().cpu())
        
        if (batch + 1) % int(ver) == 0:
            current_loss = output.detach().cpu().item()
            print(f'{(batch + 1) * batch_size}/{len(dataloader) * batch_size} | loss: {current_loss}')

            if scheduler:
                scheduler.step()
                
    all_predictions = torch.vstack(all_predictions)
    all_targets = torch.vstack(all_targets)

    train_avg_loss = loss(all_predictions, all_targets)
    train_losses.append(train_avg_loss.item())
    train_rmse.append(masked_rmse_loss(all_predictions, all_targets).item())
    print(f'Train avg loss: {train_avg_loss}')
```

```python
def test_loop(model, loss, dataloader, test_losses, test_rmse):
    model.eval()
    all_targets = []
    all_predictions = []

    is_xla = 'xla' in str(device)
    loader = pl.MpDeviceLoader(dataloader, device) if is_xla else dataloader

    with torch.no_grad():
        for batch, (X, y) in enumerate(loader):
            X, y = X.to(device), y.to(device)
    
            y_pred = model(X)
            
            output = loss(y_pred, y)
    
            all_predictions.append(y_pred.detach().cpu())
            all_targets.append(y.detach().cpu())

        all_predictions = torch.vstack(all_predictions)
        all_targets = torch.vstack(all_targets)

        if is_xla:
            torch_xla.sync()
        
        avg_loss = loss(all_predictions, all_targets)

        print(f'{"Test avg loss:":11} {avg_loss:.7f}')        
        
        mask = (all_targets != -1)
        all_predictions = all_predictions[mask]
        all_targets = all_targets[mask]

        print('Test metrics:')
        rmse = root_mean_squared_error(all_targets, all_predictions)
        print(f'RMSE: {rmse}')
        print(f'MSE: {mean_squared_error(all_targets, all_predictions)}')
        print(f'MAE: {mean_absolute_error(all_targets, all_predictions)}')
        print(f'MAPE: {mean_absolute_percentage_error(all_targets, all_predictions)}')

        test_losses.append(avg_loss.item())
        test_rmse.append(rmse)
        return avg_loss
```

## Custom model + Adam + BatchNorm + CosineAnnealingLR

```python
class CustomModel(nn.Module):
    def __init__(self, h=96, w=96):
        super().__init__()
        self.conv_stack = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.SiLU(),
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.SiLU(),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.SiLU(),
            nn.MaxPool2d(kernel_size=2)
        )
        self.flatten = nn.Flatten()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(128 * h//2 * w//2, 512),
            nn.SiLU(),
            nn.Linear(512, 512),
            nn.SiLU(),
            nn.Linear(512, 30)
        )

    def forward(self, x):
        x = self.conv_stack(x)
        x = self.flatten(x)
        x = self.linear_relu_stack(x)
        return x
```

## Визуализация слоёв (кастомной модели)

```python
def visualize_conv(layers, num_epochs, size=128, n_filters=16, lr=1e-2):
    """
    На вход подаётся часть слоёв сети, где последним слоем является свёрточный слой, который хотим визуализировать
    """
    result = {}
    
    conv = layers[-1]
    n_filters = min(conv.weight.shape[0], n_filters)

    for param in layers.parameters():
        param.requires_grad = False

    rand_idx = random.sample(range(conv.weight.shape[0]), k=n_filters)
    for i in rand_idx:
        img = torch.rand((1, 3, size, size)).to(device)
        img.requires_grad = True

        optimizer = torch.optim.Adam([img], lr=lr)

        for e in range(num_epochs):
            optimizer.zero_grad()

            output = layers(img)
            loss = -output[0][i].mean()

            loss.backward()
            optimizer.step()

        img = img.squeeze(dim=0)
        img = img.permute(1, 2, 0).cpu().detach()
        final_img = (img - img.min()) / (img.max() - img.min() + 1e-5)
        
        result[f'filter_{i}'] = final_img
        
    return result
```

## Реализация Unet

```python
class UnetConvBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.conv_stack = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1),
            nn.BatchNorm2d(out_channels),
            nn.ReLU()
        )

    def forward(self, x):
        return self.conv_stack(x)



class Unet(nn.Module):
    def __init__(self, n_classes):
        super().__init__()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        self.encoder_1 = UnetConvBlock(3, 64)
        self.encoder_2 = UnetConvBlock(64, 128)
        self.encoder_3 = UnetConvBlock(128, 256)
        self.encoder_4 = UnetConvBlock(256, 512)
        self.encoder_5 = UnetConvBlock(512, 1024)

        self.decoder_4 = UnetConvBlock(1024, 512)
        self.decoder_3 = UnetConvBlock(512, 256)
        self.decoder_2 = UnetConvBlock(256, 128)
        self.decoder_1 = UnetConvBlock(128, 64)

        self.upconv_1 = nn.ConvTranspose2d(in_channels=1024, out_channels=512, kernel_size=2, stride=2)
        self.upconv_2 = nn.ConvTranspose2d(in_channels=512, out_channels=256, kernel_size=2, stride=2)
        self.upconv_3 = nn.ConvTranspose2d(in_channels=256, out_channels=128, kernel_size=2, stride=2)
        self.upconv_4 = nn.ConvTranspose2d(in_channels=128, out_channels=64, kernel_size=2, stride=2)


        self.output_conv = nn.Conv2d(64, n_classes, kernel_size=3, padding=1)

    def forward(self, x):
        x1_encoder = self.encoder_1(x)
        x2_encoder = self.encoder_2(self.pool(x1_encoder))
        x3_encoder = self.encoder_3(self.pool(x2_encoder))
        x4_encoder = self.encoder_4(self.pool(x3_encoder))
        x5_encoder = self.encoder_5(self.pool(x4_encoder))

        def match_and_concat(encoder_tensor, upconv_tensor):
            if upconv_tensor.shape[2:] != encoder_tensor.shape[2:]:
                diff_h = encoder_tensor.size()[2] - upconv_tensor.size()[2]
                diff_w = encoder_tensor.size()[3] - upconv_tensor.size()[3]

                upconv_tensor = F.pad(upconv_tensor, [
                    diff_w // 2, diff_w - diff_w // 2,
                    diff_h // 2, diff_h - diff_h // 2
                ])
            return torch.cat([encoder_tensor, upconv_tensor], dim=1)

        x1_decoder = self.decoder_4(match_and_concat(x4_encoder, self.upconv_1(x5_encoder)))
        x2_decoder = self.decoder_3(match_and_concat(x3_encoder, self.upconv_2(x1_decoder)))
        x3_decoder = self.decoder_2(match_and_concat(x2_encoder, self.upconv_3(x2_decoder)))
        x4_decoder = self.decoder_1(match_and_concat(x1_encoder, self.upconv_4(x3_decoder)))

        output = self.output_conv(x4_decoder)

        return output
```

```python
img_size = (512, 768)

def rgb_to_labels(img):
    np_img = np.array(img)
    mask = np.zeros(np_img.shape[:2], dtype=np.int64)
    for color, label in rgb2label.items():
        mask[(np_img == color).all(axis=2)] = label
    return torch.from_numpy(mask)
    

transform = transforms.Compose([
    transforms.Resize(size=img_size, interpolation=transforms.InterpolationMode.BILINEAR),
    transforms.ToImage(),
    transforms.ToDtype(dtype=torch.float32, scale=True)
])

target_transform = transforms.Compose([
    transforms.Resize(size=img_size, interpolation=transforms.InterpolationMode.NEAREST),
    transforms.Lambda(rgb_to_labels)
])
```

```python
class AerialDroneDataset(Dataset):
    def __init__(self, images_dir, masks_dir, transform=None, target_transform=None, is_train=True, train_size=0.8,):
        self.images_dir = images_dir
        self.masks_dir = masks_dir
        self.transform = transform
        self.target_transform = target_transform


        self.images_filenames = sorted(os.listdir(images_dir))
        self.masks_filenames = sorted(os.listdir(masks_dir))

        if is_train:
            self.images_filenames = self.images_filenames[:int(len(self.images_filenames) * train_size)]
            self.masks_filenames = self.masks_filenames[:int(len(self.masks_filenames) * train_size)]
        else:
            self.images_filenames = self.images_filenames[int(len(self.images_filenames) * train_size):]
            self.masks_filenames = self.masks_filenames[int(len(self.masks_filenames) * train_size):]



    def __len__(self):
        return len(self.images_filenames)

    def __getitem__(self, idx):
        img = Image.open(self.images_dir + '/' + self.images_filenames[idx]).convert('RGB')
        mask = Image.open(self.masks_dir + '/' + self.masks_filenames[idx]).convert('RGB')

        if self.transform:
            img_transformed = self.transform(img)

        if self.target_transform:
            mask_transformed = self.target_transform(mask)

        return img_transformed, mask_transformed
```

```python
def train_loop(model, loader, loss_fn, optimizer, train_losses, batch_size, len_dataset, verbose=10):
    model.train()
    batch_losses = []


    for batch, (X, y) in enumerate(loader):
        if not is_xla:
            X, y = X.to(device), y.to(device)

        optimizer.zero_grad()
        pred = model(X)
        loss = loss_fn(pred, y)
        loss.backward()

        if is_xla:
            xm.optimizer_step(optimizer)
        else:
            optimizer.step()

        batch_losses.append(loss.item())

        if (batch + 1) % int(len_dataset/ batch_size / verbose) == 0:
            print(f'{(batch+1) * batch_size}/{len_dataset} | loss: {batch_losses[-1]:.5f}')

    avg_train_loss = sum(batch_losses) / len(batch_losses)
    print(f'Avg train loss: {avg_train_loss:.5f}')
    train_losses.append(avg_train_loss)
```

```python
def dice_coef(y_pred, y_true, num_classes):
    """
    y_pred: [N, C, H, W] - logits
    y_true: [N, H, W] - labels
    """
    dice_per_class = []
    y_pred_labels = torch.argmax(y_pred, dim=1)

    for c in range(num_classes):
        y_p = (y_pred_labels == c)
        y_t = (y_true == c)
        intersection = (y_p * y_t).sum()
        union = y_p.sum() + y_t.sum()
        
        if union != 0:
            result = 2 * intersection / union
        else:
            result = torch.tensor(1.0, device=y_pred.device)
        dice_per_class.append(result)
        
    return torch.stack(dice_per_class).mean()
```

```python
def test_loop(model, loader, loss_fn, test_losses):
    model.eval()
    batch_losses = []
    dice_coefs = []
    tp_sum = fp_sum = fn_sum = tn_sum = 0
    total_loss = 0
    num_batches = 0
    
    with torch.no_grad():
        for batch, (X, y) in enumerate(loader):
            if not is_xla:
                X, y = X.to(device), y.to(device)

            pred = model(X)
            loss = loss_fn(pred, y)
            dice = dice_coef(pred, y, num_classes=24)

            tp, fp, fn, tn = smp.metrics.get_stats(
                pred.argmax(dim=1), y.long(),
                mode='multiclass',
                num_classes=24
            )
 
            tp_sum += tp.sum(dim=0)
            fp_sum += fp.sum(dim=0)
            fn_sum += fn.sum(dim=0)
            tn_sum += tn.sum(dim=0)
            
            num_batches += 1
            total_loss += loss
            
        if is_xla:
            torch_xla.sync()

            
            
    avg_test_loss = (total_loss / num_batches).item()

    iou = smp.metrics.iou_score(tp_sum, fp_sum, fn_sum, tn_sum, reduction='macro')
    dice = smp.metrics.f1_score(tp_sum, fp_sum, fn_sum, tn_sum, reduction='macro')

    print()
    print(f'Avg test loss: {avg_test_loss:.5f}')
    print(f'Avg IoU coef: {iou:.5f}')  
    print(f'Avg dice coef: {dice:.5f}')  
    test_losses.append(avg_test_loss)
    return avg_test_loss, dice
```

```python
def label_mask_to_color_img(labels_mask, img_size):
    rgb_img = np.zeros((*img_size, 3), dtype='uint8')
    for c in range(24):
        color_mask = (labels_mask == c)
        rgb_img[color_mask] = label2rgb[c]
    return rgb_img
```
